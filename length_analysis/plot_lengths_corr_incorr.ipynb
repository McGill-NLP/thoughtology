```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
gsm = pd.read_csv("data/gsm8k.tsv", sep='\t')
math = pd.read_csv("data/math_32k.tsv", sep='\t')
aime = pd.read_csv("data/aime_50_samples.tsv", sep='\t')
```

```python
# Assuming 'gsm' is your DataFrame with 'Result' and 'Tokens' columns
plt.figure(figsize=(4, 5))

# Create the barplot with error bars (95% confidence interval by default)
sns.barplot(data=math, x="Result", y="Tokens", capsize=0.1, errorbar='se', palette=["#CB4154", "#4d6bfe"])

# Improve aesthetics
plt.xticks([0, 1], ["Incorrect", "Correct"], fontsize=12)
plt.xlabel("")
plt.ylabel("Average Token Length", fontsize=12)
plt.title("Token Length Distribution by Result", fontsize=14, fontweight="bold")

# Show grid for better readability
# plt.grid(axis="y", linestyle="--", alpha=0.3)

# Show the plot
plt.show()
```

```python
# Define the dataframes in a dictionary
datasets = {"AIME-24": aime, "MATH500": math, "GSM8k": gsm}

# Set up the figure with three subplots
fig, axes = plt.subplots(1, 3, figsize=(15, 5), sharex=True)

# Define colors for Incorrect (red) and Correct (blue/purple)
# colors = ["#4d6bfe", "#CB4154"]
colors = ["#1E5C99", "#CB4154"]

for ax, (name, df) in zip(axes, datasets.items()):
    sns.barplot(data=df, x="Result", y="Tokens", capsize=0.1, errorbar='se', palette=colors, ax=ax, order=[1, 0])
    
    ax.set_xticks([0, 1])
    ax.set_xticklabels(["Correct", "Incorrect"], fontsize=12)
    ax.set_xlabel("")
    ax.set_ylabel("Average Number of tokens", fontsize=16)
    ax.set_title(name, fontsize=16)

# Adjust layout
plt.suptitle("Average Lengths for Correct and Incorrect Thoughts", fontsize=18, y=0.94)
plt.tight_layout(rect=[0, 0.03, 1, 0.97])  # Leaves space for the suptitle

# Show the plot
plt.savefig("plots/average_token_length.pdf", dpi=300, bbox_inches='tight')
```
